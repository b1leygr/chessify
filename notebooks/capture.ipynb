```python
import cv2
cap = cv2.VideoCapture(0)
cap.set(3,1920) #width=1920
cap.set(4,1080) #height=1080
```

```python
game = input("Enter game name:")
i = 0
while True:
    img_name = input("Enter name for image (or 'q' to quit): ")
    if img_name.lower() == 'q':
        break
    if cap.isOpened():
        _,frame = cap.read()
        _,frame = cap.read()
        if _ and frame is not None:
            cv2.imwrite(f'{game}_{i}.jpg', frame)
            print(f"Image saved as {game}_{i}.jpg")
            i += 1
cap.release() #releasing camera after capturing pictures
```
